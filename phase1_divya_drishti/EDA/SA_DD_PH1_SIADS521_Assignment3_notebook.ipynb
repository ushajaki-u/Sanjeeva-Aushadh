## 1. Introduction

**The problem.** More than **33 million** Americans live with food allergies — about **1 in 10 adults** and **1 in 13 children** (FARE, 2026). But food allergens do not live only in food: they are also present inside **medications**, hidden among the "inactive" ingredients (excipients) — for example, **lactose** (a milk derivative) is one of the most common tablet fillers.

A patient with a documented food allergy can therefore be prescribed a medication whose excipients contain their allergen, because prescribing and dispensing workflows rarely cross-check inactive ingredients against an allergy profile.
```python
import json

# raw shard is gitignored (~632 MB, too big for GitHub); run once locally to profile columns
file_path = "raw_data/drug-label-0001-of-0014.json"   # relative to this notebook

with open(file_path, "r", encoding="utf-8") as f:
    data = json.load(f)

top_level_keys = set()
openfda_keys = set()

for record in data.get("results", []):
    top_level_keys.update(record.keys())
    if "openfda" in record and isinstance(record["openfda"], dict):
        openfda_keys.update(record["openfda"].keys())

print(f"Total records in file: {len(data.get('results', []))}\n")

print(f"Top-level keys ({len(top_level_keys)}):")
for key in sorted(top_level_keys):
    print(f"  - {key}")

print(f"\nopenfda sub-keys ({len(openfda_keys)}):")
for key in sorted(openfda_keys):
    print(f"  - {key}")
```

### Why I kept the columns I kept
- **Ingredients live in *more than one* field.** Active and inactive ingredients are spread across several fields (`inactive_ingredient`, `active_ingredient`, `substance_name`, `spl_product_data_elements`, `description`). So I scan **all approved ingredient fields**, not just one — missing a field could miss an allergen.

- **Route of administration.** I clean and group the `route` field (oral / injectable / topical / inhaled / other), because *how* a drug is taken strongly affects which excipients it carries — and it enables the oral-vs-injectable comparison below.

Note: I am investigating purpose, adverse reactions as we speak.

### Cleaning steps (done by the `01` / `02` pipeline)
1. **Allergen detection** — a deterministic, rule-based scan of the approved ingredient fields, tagging each of 7 allergens as **Detected / Review / Not Detected / Unknown**, with exclusion rules (e.g., "coconut milk" does not count as milk).
2. **Route cleaning** — the raw `route` is multi-valued (`DENTAL | ORAL | TOPICAL`); it is split, normalized, and turned into Yes/No group flags.

### Cleaning step, shown in code: the multi-valued `route` field

The raw `route` field is **multi-valued** — one medication can list several routes glued together, e.g. `DENTAL | ORAL | TOPICAL`. That cannot be filtered directly (`"DENTAL | ORAL | TOPICAL"` is not equal to `"ORAL"`). So we **split** it on the `|`, **normalize** each route (strip, uppercase, de-duplicate), and turn it into **Yes/No group flags** — a drug that is `ORAL | INTRAVENOUS` becomes `is_oral = True` *and* `is_injectable = True`. (This is a readable version of `clean_route()` in `01_DatasetCreation_4Status.py`.)

```python
import re
import pandas as pd

ORAL = {"ORAL", "SUBLINGUAL", "BUCCAL"}            # all administered in the mouth
TOPICAL = {"TOPICAL", "CUTANEOUS", "TRANSDERMAL", "PERCUTANEOUS"}
INHALED = {"RESPIRATORY (INHALATION)", "NASAL"}

def route_flags(raw):
    routes = sorted({p.strip().upper() for p in re.split(r"\s*\|\s*", str(raw))
                     if p.strip() and p.strip().lower() != "nan"})
    injectable = lambda r: r.startswith("INTRA") or r in {
        "INTRAVENOUS", "INTRAMUSCULAR", "SUBCUTANEOUS", "EPIDURAL", "PARENTERAL",
        "INFILTRATION", "PERINEURAL", "SOFT TISSUE", "ENDOTRACHEAL",
        "EXTRACORPOREAL", "SUBCONJUNCTIVAL"}
    return {
        "route_clean": " | ".join(routes),
        "is_oral": any(r in ORAL for r in routes),
        "is_injectable": any(injectable(r) for r in routes),
        "is_topical": any(r in TOPICAL for r in routes),
        "is_inhaled": any(r in INHALED for r in routes),
    }

# before -> after, on real route strings from the dataset
examples = ["DENTAL | ORAL | TOPICAL", "ORAL | INTRAVENOUS", "INTRAVENOUS", ""]
pd.DataFrame([{"raw route": e or "(blank)", **route_flags(e)} for e in examples])
```

### 📏 Counting rule (important)
A single medication can list **several routes of administration** - for example `ORAL | INTRAVENOUS | VAGINAL`. For the **route-level** analysis, such a drug is counted **once for each of its routes**: a drug given 4 ways counts as **4** in the route breakdown. So the per-route counts add up to more than the number of distinct medications - that is expected.

**How big is this effect here?** Small - only **114 medications (0.6%)** list 2 or more routes; 80% list none. So the exploded count is **20,165** drug-route rows, barely above the **20,000** distinct medications. We therefore report the headline on **20,000 distinct medications**, and apply the per-route counting only inside the route-level charts (the route-group flags let each drug appear in every route it belongs to).


## 4. Visualization techniques  *(rubric: 20 pts)*

I use seven chart types that build from a broad overview down to the specific evidence. For each: what it shows, why it fits here, and when you should *not* reach for it.

| Chart | What it shows | Why here | When *not* to use it |
|---|---|---|---|
| **Pie (donut)** — overall status | Every medication is exactly one of Detected / Review / Not Detected | Three mutually-exclusive slices that sum to 100% — reads the "how big is the problem" headline at a glance | More than ~4 slices, overlapping categories, or when precise comparison matters (use a bar) |
| **Bar** — prevalence per allergen | % of medications where each allergen is Detected, ranked | Allergens **overlap** (one med can carry several), so they don't sum to a whole — a bar shows each independently; a pie would mislead | Parts-of-a-whole (use pie/stacked), or trends over time (use line) |
| **Grouped bar (interactive)** — route signature | For the chosen allergen, its prevalence across the 5 route groups | The dropdown collapses 8 charts into one explorable view; a fixed shared y-axis keeps allergens honestly comparable | When you need *all* allergens × routes at once (use the heatmap) |
| **Heatmap** — allergen × route | The full matrix: % Detected for every allergen in every route, at once | The "see-everything" complement to the route signature; color encodes magnitude so patterns pop | When exact values matter more than the pattern, or the matrix is large and sparse |
| **Sunburst ⭐ (new)** — allergen → route → status | Nested rings: allergen (center) → routes → the 4-status split inside each route | Shows **hierarchy + part-to-whole at several levels at once**, and drills down — e.g. within oral-milk, how much is Detected vs Review | When precise leaf-value comparison matters (angles are hard to compare), or there are many tiny slices |
| **Waffle ⭐ (new)** — allergens per medication | A 10×10 grid, each square ≈ 1% of meds, colored by how many allergens it carries (0 / 1 / 2 / 3+) | Makes "what share of pills are clean vs multi-allergen minefields" **tangible** as 100 squares | Precise values, many categories, or continuous data (use a histogram/bar) |
| **Horizontal bar (interactive)** — worst offenders / why "Review" | The actual ingredient **words** driving each allergen's Detected (lactose, casein…) and its Review pile | Turns an abstract "Detected" tag into **auditable evidence**; the slicer inspects any allergen and surfaces the cleanup worklist | Parts-of-a-whole, or categories that aren't rankable |

### My new-to-course methods — what they reveal that a covered chart can't
- **Sunburst:** a bar shows one level and a pie shows one whole; the sunburst **nests** them — allergen → route → status in a single figure — and click-to-drill makes the *Review burden per route* visible (e.g. Alpha-Gal's route rings flooding orange). A flat chart would need several panels to say the same thing.
- **Waffle:** a histogram shows the same distribution, but the waffle renders it as a share of 100 squares, so "X% of medications carry 2+ allergens" lands viscerally for a non-expert audience.

### How the charts complement each other (and why this order)
They form a funnel from broad to specific — **scope → explore → evidence → cleanup.** The pie and bar set the headline (how big, which allergens). The route signature and heatmap answer *where* allergens hide. The sunburst drills into one allergen's route × status detail. The worst-offenders and why-"Review" bars expose the exact ingredient words — the evidence behind every tag, and the ambiguous cases still needing review. Each view hands a question to the next.

```python
# --- Chart 3 (interactive): each allergen's route-signature, via an allergen "slicer" ---
# The dropdown is an Excel-style slicer. Pick an allergen and the bars redraw to
# show THAT allergen's prevalence across the 5 route groups. Built in pure Plotly
# (graph_objects) so the interaction survives in the exported standalone HTML -
# no running server needed (a Streamlit widget would need one).
import plotly.graph_objects as go

groups = {"oral": "is_oral", "injectable": "is_injectable", "topical": "is_topical",
          "inhaled/nasal": "is_inhaled", "other-local": "is_other_route"}
ALLERGENS = {"Milk": "milk_status", "Eggs": "eggs_status", "Wheat": "wheat_status",
             "Sesame": "sesame_status", "Tree Nuts": "tree_nuts_status",
             "Peanuts": "peanuts_status", "Soy": "soy_status", "Alpha-Gal": "alpha_gal_status"}

# A route's size (its drug count) is the denominator, and it is the SAME for every
# allergen - so fix the x-axis order ONCE (biggest route first). A fixed order means
# switching allergen never reshuffles the axis, so comparisons stay honest.
route_n = {label: int(df[col].sum()) for label, col in groups.items()}
route_order = sorted(groups, key=lambda L: route_n[L], reverse=True)

# Pre-compute every allergen's per-route %, so all views can share ONE y-axis.
series = {name: [round((df[df[groups[r]]][sc] == "Detected").mean() * 100, 1) if route_n[r] else 0
                 for r in route_order]
          for name, sc in ALLERGENS.items()}
ymax = max(v for pcts in series.values() for v in pcts)   # shared, honest scale

fig = go.Figure()
for name, pcts in series.items():
    fig.add_bar(
        x=route_order, y=pcts, name=name, visible=(name == "Milk"),
        text=[f"{p}%" for p in pcts], textposition="outside", cliponaxis=False,
        marker_color="#e4572e", customdata=[route_n[r] for r in route_order],
        hovertemplate="<b>%{x}</b><br>" + name + " Detected: %{y}%"
                      "<br>%{customdata:,} drugs list this route<extra></extra>")

# The slicer: one button per allergen; each reveals only that allergen's bars.
buttons = [dict(label=name, method="update",
                args=[{"visible": [n == name for n in series]},
                      {"title.text": f"{name} prevalence by route "
                                     "(within the drugs that list each route)"}])
           for name in series]

fig.update_layout(
    title="Milk prevalence by route (within the drugs that list each route)",
    # Shared y-axis = honest cross-allergen comparison. JUDGMENT CALL: a fixed scale
    # makes tiny allergens (sesame, eggs) look flat; swap to autorange (delete the
    # range=... line) if you care more about each allergen's OWN route shape.
    yaxis=dict(title="% of that route's drugs with the allergen Detected",
               range=[0, ymax * 1.15]),
    xaxis_title="", showlegend=False, bargap=0.35,
    updatemenus=[dict(buttons=buttons, direction="down", showactive=True,
                      x=1.0, xanchor="right", y=1.20, yanchor="top",
                      bgcolor="#faf3f0", bordercolor="#e4572e", font=dict(size=13))],
    annotations=[dict(text="◀ pick an allergen", x=1.0, xref="paper", xanchor="right",
                      y=1.27, yref="paper", showarrow=False,
                      font=dict(size=12, color="#777"))])
fig.show()

# The numbers behind the chart (allergens x routes) - handy for the write-up / QA.
pd.DataFrame(series, index=route_order).T
```

### Chart 5 ⭐ NEW-TO-COURSE METHOD — Sunburst: allergen → route → status
Pick an allergen from the dropdown. The **center** is that allergen, the **middle ring** is the 5 routes it is taken by, and the **outer ring** splits each route into the 4 statuses — **Detected / Review / Not Detected / Unknown**. Not taught in SIADS 521 (satisfies the "new method" requirement), and it mirrors the interactive dashboard. Try **Alpha-Gal** — the orange *Review* ring is huge, which is the human-in-the-loop story.

```python
# --- Chart 5 (interactive): allergen -> route -> status sunburst (matches the dashboard) ---
# Dropdown picks the allergen. Center = allergen; middle ring = the 5 routes (sized by how
# many meds take each); outer ring = the 4 statuses, colored by status. Pure Plotly, so the
# dropdown survives in the exported standalone HTML - no server needed.
import plotly.express as px
import plotly.graph_objects as go

STATUS_COLORS = {"Detected": "#c1121f", "Review": "#f3722c",
                 "Not Detected": "#7fb069", "Unknown": "#b0a89f"}
STATUSES = ["Detected", "Review", "Not Detected", "Unknown"]

def _sun_rows(status_col, name):
    out = []
    for rlabel, rcol in ROUTES.items():
        sub = df[df[rcol]]
        for status in STATUSES:
            cnt = int((sub[status_col] == status).sum())
            if cnt > 0:
                out.append({"allergen": name, "route": rlabel, "status": status, "count": cnt})
    return pd.DataFrame(out)

names = list(ALLERGENS)
fig = go.Figure()
for name, status_col in ALLERGENS.items():
    sf = px.sunburst(_sun_rows(status_col, name), path=["allergen", "route", "status"],
                     values="count", color="status", color_discrete_map=STATUS_COLORS)
    tr = sf.data[0]
    tr.visible = (name == names[0])
    fig.add_trace(tr)

buttons = [dict(label=n, method="update",
                args=[{"visible": [x == n for x in names]},
                      {"title.text": f"{n}: allergen status across routes"}])
           for n in names]

fig.update_traces(textinfo="label+value", insidetextorientation="radial")
fig.update_layout(
    title=f"{names[0]}: allergen status across routes",
    margin=dict(t=90), height=560,
    updatemenus=[dict(buttons=buttons, direction="down", showactive=True,
                      x=1.0, xanchor="right", y=1.15, yanchor="top",
                      bgcolor="#faf3f0", bordercolor="#c1121f", font=dict(size=13))],
    annotations=[dict(text="◀ pick an allergen", x=1.0, xref="paper", xanchor="right",
                      y=1.22, yref="paper", showarrow=False, font=dict(size=12, color="#777"))])
fig.show()
```

### Chart 6 ⭐ NEW METHOD — Waffle: how many allergens does one medication carry?
A **waffle chart** is a 10x10 grid of 100 squares, each ~1% of medications, colored by how many allergens that medication has **Detected**. Not taught in SIADS 521. It answers *"how many pills are clean vs. multi-allergen minefields?"* at a glance. *(Detected only - "Review" is excluded, or Alpha-Gal's ubiquitous ambiguous excipients would swamp the chart.)*

```python
import numpy as np
import plotly.graph_objects as go

status_cols = [c for c in df.columns if c.endswith("_status")]
n_det = sum((df[c] == "Detected").astype(int) for c in status_cols)

def _bucket(n):
    return "0 (none)" if n == 0 else ("1" if n == 1 else ("2" if n == 2 else "3 or more"))

order = ["0 (none)", "1", "2", "3 or more"]
counts = n_det.apply(_bucket).value_counts().reindex(order, fill_value=0)
pct = counts / counts.sum() * 100

# allocate 100 squares (largest-remainder rounding)
base = np.floor(pct).astype(int)
for i in np.argsort(-(pct - base).values)[: int(100 - base.sum())]:
    base.iloc[i] += 1

colors = {"0 (none)": "#7fb069", "1": "#f4a259", "2": "#e4572e", "3 or more": "#9b2226"}
seq = [cat for cat in order for _ in range(int(base[cat]))]

fig = go.Figure(go.Scatter(
    x=[i % 10 for i in range(len(seq))], y=[i // 10 for i in range(len(seq))], mode="markers",
    marker=dict(size=24, symbol="square", color=[colors[c] for c in seq],
                line=dict(width=2, color="white")),
    text=seq, hoverinfo="text"))
for cat in order:
    fig.add_trace(go.Scatter(x=[None], y=[None], mode="markers",
        marker=dict(size=12, symbol="square", color=colors[cat]),
        name=f"{cat}: {int(counts[cat]):,} meds ({pct[cat]:.1f}%)"))
fig.update_layout(
    title="Waffle - each square = ~1% of medications, by number of allergens Detected (of 8)",
    xaxis=dict(visible=False), yaxis=dict(visible=False, scaleanchor="x"),
    plot_bgcolor="white", width=640, height=560, legend=dict(orientation="h", y=-0.08))
fig.show()
```

### Chart 7 ⭐ Worst offenders — which exact ingredients trigger each allergen? (Detected)
Pick an **allergen** (the universal filter) and **Top 3 / 5 / 10**. The bars are the *actual words* our rule matched, ranked by how many medications contain each. The default — **Milk** — is overwhelmingly **lactose** (a tablet filler), not dairy food: that's the whole thesis in one chart. Switch the slicer to see each allergen's own worst offenders (sesame → *sesame oil*, alpha-gal → *gelatin*).

```python
# --- Chart 7 (interactive): worst-offender INGREDIENTS per allergen (Detected) ---
# Two controls: the ALLERGEN slicer (our universal filter) + a Top-N selector.
# Bars = the ACTUAL words our rule matched (from {allergen}_matched_terms),
# ranked by how many medications contain each word. Pure Plotly -> works in the
# exported HTML with no server.
from collections import Counter
import plotly.graph_objects as go

ALLERGEN_COLS = {"Milk": "milk", "Eggs": "eggs", "Wheat": "wheat", "Sesame": "sesame",
                 "Tree Nuts": "tree_nuts", "Peanuts": "peanuts", "Soy": "soy", "Alpha-Gal": "alpha_gal"}

def top_terms(col, status, k=10):
    """How many medications contain each matched word, for one allergen + status."""
    c = Counter()
    for cell in df.loc[df[f"{col}_status"] == status, f"{col}_matched_terms"].dropna():
        for t in {x.strip() for x in str(cell).split("|") if x.strip()}:
            c[t] += 1
    return c.most_common(k)

detected_top = {name: top_terms(col, "Detected", 10) for name, col in ALLERGEN_COLS.items()}

def _xyt(pairs):
    return [w for w, _ in pairs], [n for _, n in pairs], [n for _, n in pairs]

x0, y0, t0 = _xyt(detected_top["Milk"])
fig = go.Figure(go.Bar(x=x0, y=y0, text=t0, textposition="outside", cliponaxis=False,
                       marker_color="#c1121f",
                       hovertemplate="<b>%{x}</b><br>%{y:,} medications<extra></extra>"))

# Control 1 - the allergen slicer (restyles the single bar trace; y rescales to it).
allergen_buttons = []
for name, pairs in detected_top.items():
    x, y, t = _xyt(pairs)
    allergen_buttons.append(dict(label=name, method="update",
        args=[{"x": [x], "y": [y], "text": [t]},
              {"yaxis.autorange": True,
               "annotations[0].text": f"{name}: most common Detected ingredients (by # medications)"}]))

# Control 2 - Top N: window the x-axis to the first N bars (they are already sorted desc).
# This persists across allergen switches, so it behaves like a sticky filter.
topn_buttons = [dict(label=f"Top {N}", method="relayout", args=[{"xaxis.range": [-0.5, N - 0.5]}])
                for N in (3, 5, 10)]

fig.update_layout(
    yaxis_title="# medications containing this word", xaxis_title="",
    showlegend=False, bargap=0.3, margin=dict(t=110),
    updatemenus=[
        dict(buttons=allergen_buttons, direction="down", showactive=True,
             x=0.0, xanchor="left", y=1.28, yanchor="top",
             bgcolor="#faf0f0", bordercolor="#c1121f", font=dict(size=13)),
        dict(buttons=topn_buttons, direction="down", showactive=True,
             x=1.0, xanchor="right", y=1.28, yanchor="top",
             bgcolor="#eef3f8", bordercolor="#577590", font=dict(size=13))],
    annotations=[
        dict(text="Milk: most common Detected ingredients (by # medications)",
             x=0.0, xanchor="left", y=1.10, yref="paper", yanchor="top",
             showarrow=False, font=dict(size=15))])
fig.show()

# Listed out (the default view, Milk) - handy readout / QA.
pd.DataFrame(detected_top["Milk"], columns=["ingredient word", "# medications"])
```

### Chart 8 ⭐ Why "Review"? — the next-cleaning worklist
**Review** means our rule caught an *ambiguous* word it couldn't confirm as the allergen. These are the words to **promote to Detected, or exclude**, in the next cleaning pass. Pick an allergen from the slicer:
- **Milk** is tiny (albumin, lactulose) — nearly nothing to fix.
- **Wheat → `starch` (3,973)** and **Eggs → `lecithin` (701)** are the classic over-flags (most starch is corn/potato; most lecithin is soy).
- **Alpha-Gal** (glycerin, magnesium stearate, stearic acid…) — the ubiquitous plant-or-animal excipients that inflate its Review rate.

```python
# --- Chart 8 (interactive): "Why Review?" - the next-cleaning worklist ---
# Review = our rule found an AMBIGUOUS word it could not confirm as the allergen.
# This ranks those words per allergen, so the next cleaning pass knows exactly what
# to resolve (promote to Detected, or add an exclusion). Same allergen slicer.
# (Reuses top_terms() + ALLERGEN_COLS defined in the Chart 7 cell above.)
review_top = {name: top_terms(col, "Review", 10) for name, col in ALLERGEN_COLS.items()}

def _xyt_rev(pairs):
    if not pairs:
        return ["— no Review items —"], [0], [""]
    return [w for w, _ in pairs], [n for _, n in pairs], [n for _, n in pairs]

rx, ry, rt = _xyt_rev(review_top["Milk"])
fig = go.Figure(go.Bar(x=rx, y=ry, text=rt, textposition="outside", cliponaxis=False,
                       marker_color="#f3722c",
                       hovertemplate="<b>%{x}</b><br>%{y:,} medications in Review<extra></extra>"))

review_buttons = []
for name, pairs in review_top.items():
    x, y, t = _xyt_rev(pairs)
    review_buttons.append(dict(label=name, method="update",
        args=[{"x": [x], "y": [y], "text": [t]},
              {"yaxis.autorange": True,
               "annotations[0].text": f'Why "Review"? - {name}: ambiguous words to resolve next pass'}]))

fig.update_layout(
    yaxis_title="# medications pushed into Review by this word", xaxis_title="",
    showlegend=False, bargap=0.3, margin=dict(t=110),
    updatemenus=[dict(buttons=review_buttons, direction="down", showactive=True,
                      x=0.0, xanchor="left", y=1.28, yanchor="top",
                      bgcolor="#fff3e6", bordercolor="#f3722c", font=dict(size=13))],
    annotations=[dict(text='Why "Review"? - Milk: ambiguous words to resolve next pass',
                      x=0.0, xanchor="left", y=1.10, yref="paper", yanchor="top",
                      showarrow=False, font=dict(size=15))])
fig.show()

# Listed out (default Milk). Switch the slicer to Wheat / Eggs / Alpha-Gal for the big piles.
pd.DataFrame(review_top["Milk"] or [("— none —", 0)],
             columns=["ambiguous word", "# meds in Review"])
```

### Exploration narrative — what I expected, what surprised me, and what only interaction revealed

**What I set out to ask.** My starting question was simple: *how widespread are food allergens in FDA-approved medications, and does it depend on how the drug is taken?* I assumed that because this is **FDA, medication-level data**, it would be clean and complete. Exploring it interactively changed that assumption fast.

**Surprise 1 — "FDA data" is far messier than I expected.** The biggest surprise was how **inconsistent the ingredient contents were**. The same ingredient shows up under many different strings — I saw **"lactose monohydrate," "anhydrous lactose," "lactose anhydrous," and "lactose"** all meaning what a human reads as the same thing, just reordered or spelled differently. To a person they're obviously identical; to a string-matching rule they're four separate tokens — which is exactly how a naive match silently misses an allergen.

**Surprise 2 — route is missing for ~80% of medications.** When I filtered by route, only about **20% of medications (≈3,977 of 20,000) have a route of administration recorded at all** — the rest are blank. That genuinely worried me: how can **FDA medication-level data be this incomplete** on something as basic as how a drug is taken? It's a data-quality finding I'd never have noticed without filtering, and it's a caveat on every route-based chart.

**The finding I could only see through interaction — Alpha-Gal's "Review" flood.** With milk I had intuition (I live with a lactose reaction), and milk is mostly **Detected** by clear words like lactose. But when I switched the allergen filter to **Alpha-Gal**, the picture flipped: it is overwhelmingly **"Review"** — about **12,500 of ~20,000 meds (≈62%)** land in Review versus only ~2,100 Detected. No single static chart showed me that; it appeared only when I changed the allergen and watched the whole dashboard reconfigure.

**What that tells me, and what's left.** I'm **not worried about "Not Detected,"** but a Review rate that high is a signal, not an answer — and I know far less about alpha-gal than about milk. It means one of two things, and I can't yet tell which: either the alpha-gal rules are **over-flagging** ambiguous excipients, **or** the ingredient evidence is sitting in **JSON fields I haven't searched yet**, so I'd have to go **record by record** to check whether other fields carry an ingredient list I missed. My next step is to **write tests** to confirm I've caught what I think I have, and to audit the Review pile field-by-field. **Fallback if hosting fails:** because the charts are Plotly, each figure exports to a **self-contained HTML** file with `fig.write_html("chart.html", include_plotlyjs="cdn")` — opens in any browser with full interactivity and no server, which the assignment accepts in place of a hosted link.
